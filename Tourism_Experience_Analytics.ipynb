### Create the first master dataframe

```python
master_df = transaction_df.merge(
    user_df,
    on="UserId",
    how="left"
)

print("Master shape:", master_df.shape)
print("Missing values after User join:")
print(master_df.isnull().sum())
```

#### Add the Item table

```python
master_df = master_df.merge(
    item_df,
    on="AttractionId",
    how="left"
)

print("Master shape:", master_df.shape)
print("Missing values after Item join:")
print(master_df.isnull().sum())
```

```python
# Merge Mode table correctly
master_df = master_df.merge(
    mode_df,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_id", "_name")
)

master_df = master_df.drop(columns=["VisitModeId"])

master_df = master_df.rename(columns={
    "VisitMode_id": "VisitModeId",
    "VisitMode_name": "VisitMode"
})

print(master_df[["VisitModeId", "VisitMode"]].head())
print("Missing VisitMode:", master_df["VisitMode"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Merge Attraction Type
master_df = master_df.merge(
    type_df,
    on="AttractionTypeId",
    how="left"
)

print(master_df[["AttractionTypeId", "AttractionType"]].head())
print("Missing AttractionType:", master_df["AttractionType"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Add User City Name
user_city_df = city_df.rename(columns={
    "CityId": "UserCityId",
    "CityName": "UserCityName"
})[["UserCityId", "UserCityName"]]

master_df = master_df.merge(
    user_city_df,
    left_on="CityId",
    right_on="UserCityId",
    how="left"
)

print(master_df[["CityId", "UserCityId", "UserCityName"]].head())
print("Missing UserCityName:", master_df["UserCityName"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Add User Country Name
user_country_df = country_df.rename(columns={
    "CountryId": "UserCountryId",
    "Country": "UserCountry"
})[["UserCountryId", "UserCountry"]]

master_df = master_df.merge(
    user_country_df,
    left_on="CountryId",
    right_on="UserCountryId",
    how="left"
)

print(master_df[["CountryId", "UserCountryId", "UserCountry"]].head())
print("Missing UserCountry:", master_df["UserCountry"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Add User Region Name
user_region_df = region_df.rename(columns={
    "RegionId": "UserRegionId",
    "Region": "UserRegion"
})[["UserRegionId", "UserRegion"]]

master_df = master_df.merge(
    user_region_df,
    left_on="RegionId",
    right_on="UserRegionId",
    how="left"
)

print(master_df[["RegionId", "UserRegionId", "UserRegion"]].head())
print("Missing UserRegion:", master_df["UserRegion"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Add User Continent Name
user_continent_df = continent_df.rename(columns={
    "ContinentId": "UserContinentId",
    "Continent": "UserContinent"
})[["UserContinentId", "UserContinent"]]

master_df = master_df.merge(
    user_continent_df,
    left_on="ContinentId",
    right_on="UserContinentId",
    how="left"
)

print(master_df[["ContinentId", "UserContinentId", "UserContinent"]].head())
print("Missing UserContinent:", master_df["UserContinent"].isnull().sum())
print("Master shape:", master_df.shape)
```

```python
# Add Attraction City Name
attraction_city_df = city_df.rename(columns={
    "CityId": "AttractionCityId",
    "CityName": "AttractionCityName"
})[["AttractionCityId", "AttractionCityName", "CountryId"]]

master_df = master_df.merge(
    attraction_city_df,
    on="AttractionCityId",
    how="left",
    suffixes=("", "_Attraction")
)

print(master_df[
    ["AttractionId", "AttractionCityId", "AttractionCityName"]
].head())

print("Missing AttractionCityName:",
      master_df["AttractionCityName"].isnull().sum())

print("Master shape:", master_df.shape)
```

```python
# Add Attraction Country Name
attraction_country_df = country_df.rename(columns={
    "CountryId": "AttractionCountryId",
    "Country": "AttractionCountry"
})[["AttractionCountryId", "AttractionCountry"]]

master_df = master_df.merge(
    attraction_country_df,
    left_on="CountryId_Attraction",
    right_on="AttractionCountryId",
    how="left"
)

print(master_df[
    ["Attraction", "AttractionCityName",
     "AttractionCountryId", "AttractionCountry"]
].head())

print("Missing AttractionCountry:",
      master_df["AttractionCountry"].isnull().sum())

print("Master shape:", master_df.shape)
```

#### Average Rating by Attraction

```python
# Calculate average rating and visit count
attraction_rating = (
    master_df.groupby("Attraction")
    .agg(
        Average_Rating=("Rating", "mean"),
        Visit_Count=("Rating", "count")
    )
    .sort_values(
        ["Average_Rating", "Visit_Count"],
        ascending=[False, False]
    )
)

# Top 10
top_rated_attractions = attraction_rating.head(10)

# Output
print("Top 10 Attractions by Average Rating:")
print(top_rated_attractions.round(2))

# Visualization
plt.figure(figsize=(10, 6))

plt.barh(
    top_rated_attractions.index[::-1],
    top_rated_attractions["Average_Rating"].values[::-1]
)

plt.title("Top 10 Attractions by Average Rating")
plt.xlabel("Average Rating")
plt.ylabel("Attraction")
plt.xlim(0, 5)

plt.tight_layout()
plt.show()
```

Mount Semeru Volcano has the highest average rating at 4.70, but it has only 83 visits. This means a high average rating does not necessarily mean an attraction is popular.

Waterbom Bali stands out strongly because it has both a very high average rating of 4.65 and a large 6,429 visits. This makes it one of the strongest attractions when considering both popularity and satisfaction.

## Feature engineering and preprocessing for ML

### Regression

```python
regression_features = [
    "VisitYear",
    "VisitMonth",
    "VisitMode",
    "Attraction",
    "AttractionType",
    "AttractionCity",
    "UserCountry",
    "UserRegion",
    "UserContinent"
]

X = master_df[regression_features].copy()
y = master_df["Rating"].copy()

print("Feature Dataset Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nSelected Features:")
print(X.columns.tolist())

print("\nData Types:")
print(X.dtypes)

print("\nMissing Values:")
print(X.isnull().sum())

print("\nTarget Summary:")
print(y.describe())
```

#### Encode Categorical Features

```python
categorical_columns = [
    "VisitMode",
    "Attraction",
    "AttractionType",
    "AttractionCity",
    "UserCountry",
    "UserRegion",
    "UserContinent"
]

X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("Shape Before Encoding:", X.shape)
print("Shape After Encoding :", X_encoded.shape)

print("\nTotal Features After Encoding:")
print(X_encoded.shape[1])

print("\nData Types:")
print(X_encoded.dtypes.value_counts())

print("\nMissing Values:")
print(X_encoded.isnull().sum().sum())

print("\nFirst 10 Encoded Columns:")
print(X_encoded.columns[:10].tolist())
```

### Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Feature Shape:", X_train.shape)
print("Testing Feature Shape :", X_test.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape :", y_test.shape)

print("\nTraining Rating Mean:", round(y_train.mean(), 4))
print("Testing Rating Mean :", round(y_test.mean(), 4))

print("\nTraining Rating Distribution:")
print(
    y_train.value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nTesting Rating Distribution:")
print(
    y_test.value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

##### Baseline Regression Model

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

baseline_model = DummyRegressor(strategy="mean")

baseline_model.fit(X_train, y_train)

baseline_pred = baseline_model.predict(X_test)

baseline_mae = mean_absolute_error(y_test, baseline_pred)
baseline_mse = mean_squared_error(y_test, baseline_pred)
baseline_rmse = np.sqrt(baseline_mse)
baseline_r2 = r2_score(y_test, baseline_pred)

print("Baseline Regression Results")
print("---------------------------")
print("MAE :", round(baseline_mae, 4))
print("MSE :", round(baseline_mse, 4))
print("RMSE:", round(baseline_rmse, 4))
print("R²  :", round(baseline_r2, 4))

print("\nBaseline Prediction:")
print(round(baseline_pred[0], 4))
```

### Compare Multiple Regression Models

```python
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor
)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import pandas as pd
import numpy as np


models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(alpha=1.0),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=10,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=15,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=100,
        max_depth=15,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    "HistGradientBoosting": HistGradientBoostingRegressor(
        learning_rate=0.05,
        max_iter=200,
        max_leaf_nodes=31,
        random_state=42
    )
}


results = []

for name, model in models.items():

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    })


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="R2",
    ascending=False
).reset_index(drop=True)

results_df[["MAE", "MSE", "RMSE", "R2"]] = (
    results_df[["MAE", "MSE", "RMSE", "R2"]]
    .round(4)
)

print("Regression Model Comparison")
print("---------------------------")
print(results_df.to_string(index=False))

print("\nBest Initial Model:")
print(results_df.iloc[0])
```

```python
# Build Attraction Popularity Score
recommendation_df = (
    master_df
    .groupby([
        "AttractionId",
        "Attraction",
        "AttractionType",
        "AttractionCity"
    ])
    .agg(
        VisitCount=("TransactionId", "count"),
        UniqueVisitors=("UserId", "nunique"),
        AverageRating=("Rating", "mean")
    )
    .reset_index()
)

recommendation_df["AverageRating"] = (
    recommendation_df["AverageRating"].round(2)
)

recommendation_df = recommendation_df.sort_values(
    ["VisitCount", "AverageRating"],
    ascending=[False, False]
).reset_index(drop=True)

print("ATTRACTION POPULARITY DATA")
print("--------------------------")

print(
    recommendation_df[
        [
            "Attraction",
            "AttractionType",
            "AttractionCity",
            "VisitCount",
            "UniqueVisitors",
            "AverageRating"
        ]
    ].head(10).to_string(index=False)
)
```

```python
# Create Weighted Recommendation Score
C = master_df["Rating"].mean()

m = recommendation_df["VisitCount"].quantile(0.60)

recommendation_df["RecommendationScore"] = (
    (
        recommendation_df["VisitCount"]
        / (recommendation_df["VisitCount"] + m)
    )
    * recommendation_df["AverageRating"]
    +
    (
        m
        / (recommendation_df["VisitCount"] + m)
    )
    * C
)

recommendation_df["RecommendationScore"] = (
    recommendation_df["RecommendationScore"].round(3)
)

recommendation_ranked = recommendation_df.sort_values(
    "RecommendationScore",
    ascending=False
).reset_index(drop=True)

print("Global Average Rating:", round(C, 3))
print("Minimum Visit Threshold:", round(m, 2))

print("\nTOP 10 RECOMMENDED ATTRACTIONS")
print("------------------------------")

print(
    recommendation_ranked[
        [
            "Attraction",
            "AttractionType",
            "AttractionCity",
            "VisitCount",
            "AverageRating",
            "RecommendationScore"
        ]
    ].head(10).to_string(index=False)
)
```

```python
# Build Content-Based Recommendation Features
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics.pairwise import cosine_similarity

attraction_profiles = (
    master_df[
        [
            "AttractionId",
            "Attraction",
            "AttractionType",
            "AttractionCity"
        ]
    ]
    .drop_duplicates("AttractionId")
    .sort_values("AttractionId")
    .reset_index(drop=True)
)

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

content_matrix = encoder.fit_transform(
    attraction_profiles[
        ["AttractionType", "AttractionCity"]
    ]
)

similarity_matrix = cosine_similarity(
    content_matrix
)

print("ATTRACTION PROFILE DATA")
print("-----------------------")
print("Number of Attractions :", len(attraction_profiles))
print("Content Matrix Shape  :", content_matrix.shape)
print("Similarity Matrix     :", similarity_matrix.shape)

print("\nSample Attraction Profiles")
print("--------------------------")
print(
    attraction_profiles.head(10).to_string(index=False)
)
```

```python
# Test Top 5 Similar Attractions
def recommend_similar_attractions(attraction_name, top_n=5):
    if attraction_name not in attraction_profiles["Attraction"].values:
        return "Attraction not found."

    idx = attraction_profiles[
        attraction_profiles["Attraction"] == attraction_name
    ].index[0]

    similarity_scores = list(
        enumerate(similarity_matrix[idx])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = similarity_scores[1:top_n + 1]

    recommended_indices = [
        i[0] for i in similarity_scores
    ]

    result = attraction_profiles.loc[
        recommended_indices,
        [
            "Attraction",
            "AttractionType",
            "AttractionCity"
        ]
    ].copy()

    result["SimilarityScore"] = [
        round(i[1], 3)
        for i in similarity_scores
    ]

    return result


test_recommendations = recommend_similar_attractions(
    "Kuta Beach - Bali",
    top_n=5
)

print("SIMILAR ATTRACTIONS")
print("-------------------")
print(test_recommendations.to_string(index=False))
```

```python
# Build Hybrid Recommendation System
from sklearn.preprocessing import MinMaxScaler

score_scaler = MinMaxScaler()

recommendation_ranked["NormalizedRecommendationScore"] = (
    score_scaler.fit_transform(
        recommendation_ranked[["RecommendationScore"]]
    )
)

attraction_hybrid = attraction_profiles.merge(
    recommendation_ranked[
        [
            "AttractionId",
            "VisitCount",
            "AverageRating",
            "NormalizedRecommendationScore"
        ]
    ],
    on="AttractionId",
    how="left"
)


def hybrid_recommendations(attraction_name, top_n=5):

    if attraction_name not in attraction_profiles["Attraction"].values:
        return "Attraction not found."

    idx = attraction_profiles[
        attraction_profiles["Attraction"] == attraction_name
    ].index[0]

    results = attraction_hybrid.copy()

    results["SimilarityScore"] = similarity_matrix[idx]

    results = results[
        results["Attraction"] != attraction_name
    ].copy()

    results["HybridScore"] = (
        0.70 * results["SimilarityScore"]
        + 0.30 * results["NormalizedRecommendationScore"]
    )

    results = results.sort_values(
        "HybridScore",
        ascending=False
    ).head(top_n)

    return results[
        [
            "Attraction",
            "AttractionType",
            "AttractionCity",
            "VisitCount",
            "AverageRating",
            "SimilarityScore",
            "HybridScore"
        ]
    ]


hybrid_result = hybrid_recommendations(
    "Kuta Beach - Bali",
    top_n=5
)

print("HYBRID RECOMMENDATIONS")
print("----------------------")
print(
    hybrid_result.to_string(
        index=False,
        formatters={
            "SimilarityScore": "{:.3f}".format,
            "HybridScore": "{:.3f}".format
        }
    )
)
```
